# KidneyCare-AI: 04 - Benchmark Clinical Model Comparison

Comprehensive model comparison across classical machine learning (Logistic Regression, Random Forest, SVM, Gradient Boosting) and deep learning (Tabular MLP, Voting Ensemble) on the held-out clinical test set.

### Clinical Metrics Evaluated:
- **Sensitivity (Recall)**: Rate of identifying true CKD cases (crucial for screening).
- **Specificity**: True negative rate among non-CKD controls.
- **Brier Score & Expected Calibration Error (ECE)**: Probability calibration reliability.
- **Macro F1-Score & ROC-AUC**: Balanced discriminative ability.

```python
import pandas as pd
from pathlib import Path

report_path = Path("../reports/evaluation/model_benchmark_report.csv")
if not report_path.exists():
    report_path = Path("../experiments/experiment_registry.csv")

if report_path.exists():
    df_comp = pd.read_csv(report_path)
    print("=== Clinical Benchmark Model Evaluation ===")
    cols_to_show = [c for c in ["model_type", "accuracy", "sensitivity", "specificity", "f1_macro", "roc_auc", "brier_score", "latency_ms"] if c in df_comp.columns]
    print(df_comp[cols_to_show].to_string(index=False))
else:
    print("Run 'python3 scripts/evaluate_all.py' to generate benchmark results table.")
```
