# KidneyCare-AI: 01 - Exploratory Clinical Data Analysis (EDA)

Exploratory analysis of structured clinical biomarkers for Chronic Kidney Disease (CKD) detection.

### Clinical Context:
- **Dataset**: UCI Chronic Kidney Disease (Apollo Hospitals cohort: 400 patients, 24 clinical attributes).
- **Key Biomarkers**: Serum Creatinine (`sc`), Blood Urea (`bu`), Hemoglobin (`hemo`), Albumin (`al`), Specific Gravity (`sg`), Blood Glucose Random (`bgr`).
- **Target Definition**: `ckd` (Chronic Kidney Disease) vs `notckd` (Non-CKD control).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from kidneycare.data.dataset_downloader import DatasetDownloader

# Ensure clinical dataset is present or generate synthetic baseline
downloader = DatasetDownloader(raw_dir="../data/raw/clinical")
data_path = downloader.acquire(mode="synthetic")
df = pd.read_csv(data_path)
print(f"Dataset Dimensions: {df.shape} (Patients x Features)")
df.head()
```

```python
# Inspect target distribution
plt.figure(figsize=(7, 4))
target_col = "classification" if "classification" in df.columns else "target"
counts = df[target_col].value_counts()
plt.bar([str(k) for k in counts.index], counts.values, color=["#d95f02", "#1b9e77"], edgecolor="black")
plt.title("CKD vs Control Cohort Distribution", fontsize=13, fontweight="bold")
plt.xlabel("Clinical Status")
plt.ylabel("Number of Patients")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Profile primary renal biomarkers: Serum Creatinine and Hemoglobin
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Serum Creatinine (Normal: 0.6 - 1.2 mg/dL)
if "sc" in df.columns:
    sns.histplot(data=df, x="sc", hue=target_col, kde=True, ax=axes[0], palette="Set1")
    axes[0].axvline(1.2, color="red", linestyle="--", label="Upper Normal (1.2 mg/dL)")
    axes[0].set_title("Serum Creatinine Distribution (mg/dL)")
    axes[0].legend()

# Hemoglobin (Normal: 12 - 17.5 g/dL)
if "hemo" in df.columns:
    sns.histplot(data=df, x="hemo", hue=target_col, kde=True, ax=axes[1], palette="Set1")
    axes[1].axvline(12.0, color="red", linestyle="--", label="Anemia Threshold (12 g/dL)")
    axes[1].set_title("Hemoglobin Distribution (g/dL)")
    axes[1].legend()

plt.tight_layout()
plt.show()
```
