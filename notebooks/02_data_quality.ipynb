# KidneyCare-AI: 02 - Clinical Data Quality & Hygiene Audit

Rigorous verification of clinical data hygiene, missing lab test patterns, plausible physiological ranges, and duplicate records.

### Key Requirements:
1. Detect and quantify missing laboratory values across clinical features without silent deletion.
2. Identify physiological range violations (e.g., negative age, impossible blood pressure).
3. Verify patient deduplication and identifier integrity.
4. Profile dataset memory footprint and generate structured quality reports.

```python
import pandas as pd
import numpy as np
from pathlib import Path
from kidneycare.data.data_validator import DataValidator
from kidneycare.data.data_quality import DataQualityAuditor

data_path = Path("../data/raw/clinical/synthetic_ckd.csv")
if not data_path.exists():
    data_path = Path("../data/raw/clinical/chronic_kidney_disease.csv")

df = pd.read_csv(data_path)
auditor = DataQualityAuditor()
report = auditor.generate_report(df)

print("=== Clinical Data Quality Report ===")
print(f"Total Patient Records: {report.total_records}")
print(f"Total Clinical Features: {report.total_features}")
print(f"Duplicate Records: {report.duplicate_rows}")
print(f"Overall Missing Cell %: {report.missing_percentage:.2f}%")
print(f"Memory Footprint: {report.memory_usage_mb:.3f} MB")
```

```python
# Feature-level missingness breakdown
missing_by_col = df.isnull().sum()[df.isnull().sum() > 0].sort_values(ascending=False)
if len(missing_by_col) > 0:
    print("Features with missing values:")
    print(missing_by_col)
else:
    print("No missing values in current preprocessed view.")
```
