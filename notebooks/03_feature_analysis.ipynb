# KidneyCare-AI: 03 - Clinical Feature Engineering & Biomarker Interactions

Domain-driven renal feature engineering following KDIGO guidelines.

### Clinically Formulated Features:
1. **eGFR (Estimated Glomerular Filtration Rate)** via 4-variable MDRD equation:
   $$\text{eGFR} = 175 \times (\text{SCr})^{-1.154} \times (\text{Age})^{-0.203} \times [0.742 \text{ if female}]$$
2. **BUN/Creatinine Ratio**: Distinguishes prerenal azotemia ($>20:1$) from acute tubular necrosis / intrinsic renal damage ($<15:1$).
3. **Electrolyte Gap**: $\text{Na}^+ - \text{Cl}^-$ balance for metabolic acidosis assessment.
4. **Hypertension-Diabetes Interaction**: Synergistic cardiovascular-renal risk factor.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from kidneycare.data.feature_engineering import ClinicalFeatureEngineer

data_path = Path("../data/raw/clinical/synthetic_ckd.csv")
if not data_path.exists():
    data_path = Path("../data/raw/clinical/chronic_kidney_disease.csv")

df = pd.read_csv(data_path)
fe = ClinicalFeatureEngineer()
df_engineered = fe.fit_transform(df)

print("Engineered Features:", [c for c in df_engineered.columns if c not in df.columns])
df_engineered[["sc", "bu", "bun_cr_ratio", "electrolyte_gap", "egfr_mdrd"]].head()
```

```python
# Visualize eGFR by target classification
target_col = "classification" if "classification" in df_engineered.columns else "target"
plt.figure(figsize=(8, 4))
sns.boxplot(data=df_engineered, x=target_col, y="egfr_mdrd", palette="Set2")
plt.axhline(60, color="red", linestyle="--", label="KDIGO Stage 3 CKD Threshold (60 mL/min/1.73m²)")
plt.title("Estimated GFR (MDRD) by Disease Status", fontsize=13, fontweight="bold")
plt.ylabel("eGFR (mL/min/1.73m²)")
plt.legend()
plt.tight_layout()
plt.show()
```
